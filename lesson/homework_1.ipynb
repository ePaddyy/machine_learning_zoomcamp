```python
df_asia = df[df["origin"] == "Asia"]
X = df_asia[["vehicle_weight", "model_year"]].head(7)

XT = X.transpose()
XTX = XT.dot(X)

inv = np.linalg.inv(XTX)
inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = inv.dot(XT.dot(y))
print(w)
w.sum()
```
